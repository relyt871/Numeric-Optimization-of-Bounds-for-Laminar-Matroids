# Numerical Optimization for the Laminar Matroid Lower Bound

In the paper, we describe a recursive construction that, given parameters $r_1>r_2>\ldots>r_\ell$, recursively constructs an instance $L_\ell$, such that
- $\mathrm{OPT}(L_\ell) \to 2+\sum_{i=1}^\ell r_i$;
- the expected value of the best non-adaptive algorithm can be upper bounded by $\mathrm{ALG}_{\ell}(\ell+1)$, where $\mathrm{ALG}_\ell(s)$ denotes an upper bound on the algorithm's value when at most $s$ items can be selected from $L_\ell$, and can be recursively evaluated as:
$$
\mathrm{ALG}_\ell(s) := \sup_{\lambda\geq 0}\left\{r_\ell\mathbb{E}[\min\{Z_\lambda,s\}] + \mathbb{E}\left[\mathrm{ALG}_{\ell-1}(\min\{\ell,(s-Z_\lambda)^+\})\right]\right\} ~,
$$
where $Z_\lambda\sim\mathrm{Pois}(\lambda)$.


## Numeric Estimate of the Supremum over $\lambda$

For fixed $\ell$, $s$, and $r_\ell$, define
$$
F_{\ell,s}(\lambda) = r_\ell \mathbb{E}[\min\{Z_\lambda,s\}] + \mathbb{E}\left[\mathrm{ALG}_{\ell-1}(\min\{\ell,(s-Z_\lambda)^+\})\right] ~.
$$
We evaluate an upper bound of the supremum over $\lambda\geq 0$ by first evaluating on a finite grid and then bounding the error using Lipschitz continuity.


### Lipschitz Continuity
Write $F_{\ell,s}(\lambda)=\mathbb{E}[g(Z_\lambda)] = \sum_{z\geq 0}g(z)\Pr[Z_\lambda=z]$, where
$$
g(z) = r_\ell\min\{z,s\} + \mathrm{ALG}_{\ell-1}(\min\{\ell,(s-z)^+\}) ~.
$$
Poisson variables have the following property (assuming $\Pr[Z_\lambda = -1]=0$):
$$
\frac{d}{d\lambda}\Pr[Z_\lambda=z] = \Pr[Z_\lambda = z-1] - \Pr[Z_\lambda=z] ~,
$$
therefore
$$
\frac{d}{d\lambda}\mathbb{E}[g(Z_\lambda)] = \sum_{z\geq 0}g(z)(\Pr[Z_\lambda=z-1] - \Pr[Z_\lambda = z]) =  \mathbb{E}[g(Z_\lambda + 1) - g(Z_\lambda)] ~.
$$
Let $M_{\ell,s} = \max_{z\geq 0}|g(z+1)-g(z)|$, then $F_{\ell,s}$ is $M_{\ell,s}$-Lipschitz.
Since $g(z)$ is constant for $z\geq s$, this maximum can be computed by taking maximum over $0\leq z\leq s$.

### Numeric Computation
Next we go through the numeric details.

Suppose for every $0\leq s\leq \ell$ we have an estimate $\widehat{\mathrm{ALG}}_{\ell-1}(s)$ for each $\mathrm{ALG}_{\ell-1}(s)$ such that
$$
\mathrm{ALG}_{\ell-1}(s)\leq \widehat{\mathrm{ALG}}_{\ell-1}(s)\leq \mathrm{ALG}_{\ell-1}(s) + h_{\ell-1} ~.
$$
Then we can compute
$$
\hat g(z) = r_\ell\min\{z,s\} + \widehat{\mathrm{ALG}}_{\ell-1}(\min\{\ell,(s-z)^+\}) ~,
$$
which satisfies $g(z)\leq \hat g(z)\leq g(z)+h_{\ell-1}$. Then we know that $F_{\ell,s}(\cdot)$ is $\hat M_{\ell,s}$-Lipschitz for
$$
\hat M_{\ell,s} = \max_{0\leq z\leq s}|\hat g(z+1)-\hat g(z)| + 2h_{\ell-1} ~.
$$

If we compute the maximum of $\hat F_{\ell,s}(\lambda) = \mathbb{E}[\hat g(Z_\lambda)]$ over a finite grid $G_{\ell,x} = \{kx_{\ell,s}, k\in \mathbb{Z}_{\geq 0}\}$ where $x_{\ell,s}$ is a small enough positive real number, then since $\hat F_{\ell,s}(\lambda) \geq F_{\ell,s}(\lambda)$ and $F$ is $\hat M_{\ell,s}$-Lipschitz, we have
$$
\sup_{\lambda \geq 0}F_{\ell,s}(\lambda) \leq \max_{\lambda\in G_{\ell,s}}F_{\ell,s}(\lambda) + \frac{\hat M_{\ell,s} x_{\ell,s}}{2} \leq \max_{\lambda\in G_{\ell,s}}\hat F_{\ell,s}(\lambda)+\frac{\hat M_{\ell,s}x_{\ell,s}}{2} ~.
$$

### Bounding the Useful Range of $\lambda$
Of course, we cannot enumerate all $\lambda\in G_x$. We next show that it is enough to evaluate $\lambda$ on a bounded range.

Since $g(z)$ is constant for $z\geq s$, for every $\lambda$ we have
$$
\hat F_{\ell,s}(\lambda) = \mathbb{E}[\hat g(Z_\lambda)] \leq \hat g(s)+\hat D_{\ell,s}\Pr[Z_\lambda<s] ~,
$$
where $\hat D_{\ell,s}=\max_{0\leq z<s}(\hat g(z)-\hat g(s))^+$. Since $\Pr[Z_\lambda<s]$ converges to $0$ as $\lambda\to\infty$, $\hat F_{\ell,s}(\lambda)$ must also converge to $\hat g(s)$.

In the computation, we choose $\Lambda\in G_x$ large enough so that
$$
\hat D_{\ell,s}\Pr[Z_\Lambda<s] \leq \frac{\hat M_{\ell,s}x_{\ell,s}}{2} ~.
$$
Then for all $\lambda \geq \Lambda$, we have $\hat F_{\ell,s}(\lambda)\leq \hat g(s) + \frac{\hat M_{\ell,s}x_{\ell,s}}{2}$.

Therefore, we conclude that the following is an upper bound on $\mathrm{ALG}_\ell(s)$:
$$
\widehat{\mathrm{ALG}}_{\ell}(s):=\max\left\{
\max_{\lambda\in G_x\cap[0,\Lambda]}\hat F_{\ell,s}(\lambda) + \frac{\hat M_{\ell,s}x_{\ell,s}}{2},
\hat g(s)+\frac{\hat M_{\ell,s}x_{\ell,s}}{2}
\right\} ~.
$$

Since $g(z)\leq \hat g(z)\leq g(z)+h_{\ell-1}$, we have
$$
F_{\ell,s}(\lambda) \leq \hat F_{\ell,s}(\lambda) \leq F_{\ell,s}(\lambda)+h_{\ell-1} ~,
$$
Hence the grid-based computation overestimates by at most $h_\ell = h_{\ell-1} + \max_{s=0}^{\ell+1}\frac{\hat M_{\ell,s}x_{\ell,s}}{2}$, i.e., for all $0\leq s \leq \ell+1$,
$$
\mathrm{ALG}_{\ell}(s) \leq \widehat{\mathrm{ALG}}_{\ell}(s) \leq \mathrm{ALG}_{\ell}(s) + h_\ell ~.
$$

## Code

The following is the code that implements the numeric estimation described above for a fixed vector of $(r_1,r_2,\ldots,r_L)$.

```python
import numpy as np
from scipy.special import gammaincc


L = 3
MAX_S = L + 1
r_step = 0.01
lambda_grid_step = 0.001
max_lambda = 25.0

def preprocessing(max_s=MAX_S, grid_step=lambda_grid_step, max_lambda=max_lambda):
    """Preprocess the Poisson weights needed in the computation"""
    lambda_values = np.arange(0.0, max_lambda + 0.5 * grid_step, grid_step)
    weights = {}
    cdf_less = {}

    for s in range(max_s + 1):
        if s == 0:
            weights[s] = np.ones((len(lambda_values), 1))
            cdf_less[s] = np.zeros(len(lambda_values))
            continue

        columns = []
        previous_cdf = np.zeros(len(lambda_values))
        for z in range(s):
            current_cdf = gammaincc(z + 1, lambda_values)  # Pr[Z_lambda <= z]
            columns.append(np.clip(current_cdf - previous_cdf, 0.0, 1.0))
            previous_cdf = current_cdf

        columns.append(np.clip(1.0 - previous_cdf, 0.0, 1.0))  # Pr[Z_lambda >= s]
        weights[s] = np.vstack(columns).T
        cdf_less[s] = np.clip(previous_cdf, 0.0, 1.0)  # Pr[Z_lambda < s]

    return lambda_values, weights, cdf_less


lambda_values, poisson_weights, poisson_cdf_less = preprocessing()


def compute_estimates(r_values):
    """Compute certified upper estimates for a fixed vector (r_1,...,r_L)."""
    alg_hat = np.zeros((L + 1, MAX_S + 1))
    error = np.zeros((L + 1, MAX_S + 1))

    alg_hat[0, 1:] = 1.0

    for ell in range(1, L + 1):
        r_ell = r_values[ell - 1]
        h_prev = float(np.max(error[ell - 1, : ell + 1]))

        for s in range(1, ell + 2):
            hat_g = np.zeros(s + 1)
            for z in range(s + 1):
                residual = min(ell, max(s - z, 0))
                hat_g[z] = r_ell * min(z, s) + alg_hat[ell - 1, residual]

            max_difference = float(np.max(np.abs(np.diff(hat_g))))
            hat_M = max_difference + 2.0 * h_prev
            discretization_error = 0.5 * hat_M * lambda_grid_step

            hat_D = max(0.0, float(np.max(hat_g[:s]) - hat_g[s]))
            if hat_D == 0.0:
                lambda_index = 0
            else:
                admissible = np.flatnonzero(hat_D * poisson_cdf_less[s] <= discretization_error)
                if len(admissible) == 0:
                    raise ValueError("Increase max_lambda: the tail condition was not reached.")
                lambda_index = int(admissible[0])

            grid_values = poisson_weights[s][: lambda_index + 1].dot(hat_g)
            alg_hat[ell, s] = max(float(np.max(grid_values)), float(hat_g[s])) + discretization_error
            error[ell, s] = h_prev + discretization_error

    opt_values = [2.0 + sum(r_values[:ell]) for ell in range(L + 1)]
    final_alg = float(alg_hat[L, L + 1])
    final_opt = opt_values[L]

    return {
        "r_values": tuple(r_values),
        "alg_hat": alg_hat,
        "error": error,
        "opt_values": opt_values,
        "ALG_upper": final_alg,
        "OPT": final_opt,
        "ratio": final_opt / final_alg,
    }
```

Finally, we perform a grid search over vectors of $(r_1,r_2,r_3)$.

```python
def grid_search():
    """Grid search over 0 < r_3 <= r_2 <= r_1 <= 0.5 with step size r_step."""
    grid_size = int(round(0.5 / r_step))
    r_grid = [round(i * r_step, 10) for i in range(1, grid_size + 1)]

    best = None
    for i1, r1 in enumerate(r_grid):
        for i2 in range(i1 + 1):
            r2 = r_grid[i2]
            for i3 in range(i2 + 1):
                r3 = r_grid[i3]
                result = compute_estimates((r1, r2, r3))
                if best is None or result["ratio"] > best["ratio"]:
                    best = result
    return best


best = grid_search()
print("best r-vector:", best["r_values"])
print("OPT(L_3):", best["OPT"])
print("certified upper estimate of ALG_3(4):", best["ALG_upper"])
print("certified lower-bound ratio:", best["ratio"])
```
